#### 15. Create a 2d array with 1 on the border and 0 inside (★☆☆)

````python
hint(15)

Z = np.ones((10,10))
Z1 = np.ones((10,10))
Z2 = np.ones((10,10))

# only set the first and last column to be 1
Z1[:,1:-1] = 0
print(Z1)

# only set the first and last row to be 1
Z2[[2,4],:] = 0
print(Z2)

# set both the first and last column and row to be 1
Z[1:-1,1:-1] = 0
print(Z)

answer(15)
````

#### 16. How to add a border (filled with 0's) around an existing array? (★☆☆)

#### 27. Consider an integer vector Z, which of these expressions are legal? (★☆☆)
```python
Z**Z
2 << Z >> 2
Z <- Z
1j*Z
Z/1/1
Z<Z>Z
```

````python
import numpy as np

Z = np.array([1, 2, 3, 4, 5])
print(Z)

# [1^1, 2^2, 3^3, 4^4]
print(Z**Z)

# bit shifting
print(2 << Z >> 2)

# Comparison between Z and -Z
print(Z <- Z)

# Complex multiplication
print(1j*Z)

# Division
print(Z/2/2)

#print(Z<Z>)
````

#### 38. Consider a generator function that generates 10 integers and use it to build an array (★☆☆)

````python
answer(38)

def generate():
    for i in range(10):
        yield i

lst = list(generate())
print(lst)

# iterable: the generator generate()
# count: how many items to read; -1 means read all items until the iterable is exhausted
Z = np.fromiter(generate(),dtype=float,count=-1)
print(Z)
````
